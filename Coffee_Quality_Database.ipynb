## STEP 4 — DATA CLEANING

```python
# Create a copy for final cleaning 

df_clean=df_processed.copy()

# Removing columns that are not useful for prediction

columns_to_drop=[
    'Unnamed: 0',
    'Species','Lot.Number',
    'ICO.Number','Farm.Name',
    'Mill','Owner','Owner.1',
    'Certification.Address',
    'Certification.Contact',
    'Producer','Company',
    'Altitude'
]

df_clean=df_clean.drop(columns=columns_to_drop)

print('Shape after dropping unncessary columns:', df_clean.shape)
print(df_clean.columns.tolist())
```

```python
# Clean Bag.Weight
# Bag.Weight → object
# Converting this to number

df_clean['Bag.Weight']=(
    df_clean['Bag.Weight']
    .astype(str)
    .str.extract(r'(\d+\.?\d*)')[0]
)

print(df_clean['Bag.Weight'].head())
print(df_clean['Bag.Weight'].dtype)
```

```python
df_clean['Bag.Weight'] = pd.to_numeric(
    df_clean['Bag.Weight'],
    errors='coerce'
)

print(df_clean['Bag.Weight'].head())
print(df_clean['Bag.Weight'].dtype)
print("Missing Bag.Weight:", df_clean['Bag.Weight'].isna().sum())
```

```python
# Clean Harvest.Year

print(df_clean['Harvest.Year'].value_counts(dropna=False).head(30))
```

- We can see above Harvest.Year is messy text, so we should not simply use pd.to_numeric() on it.
- for example:
2012 → 2012
2013/2014 → 2013
2017 / 2018 → 2017
2010-2011 → 2010
March 2010 → 2010
January 2011 → 2011
08/09 crop → likely 2008 based on the text
TEST, mmm, Mayo a Julio → cannot reliably determine a year → NaN

```python
# Convert Harvest.Year to a numeric year

df_clean['Harvest.Year']=(
    df_clean['Harvest.Year']
    .astype(str)
    .str.extract(r'((?:19|20)\d{2})')[0]
)

df_clean['Harvest.Year']=pd.to_numeric(
    df_clean['Harvest.Year'],
    errors='coerce'
)

print(df_clean['Harvest.Year'].value_counts(dropna=False).sort_index())
print()
print('Data Type:',df_clean['Harvest.Year'].dtype)
print('Missing Values:',df_clean['Harvest.Year'].isna().sum())
```

```python
# Handle missing values

missing=df_clean.isnull().sum()
print(missing[missing>0].sort_values(ascending=False))
```

```python
# Categorical columns
categorical_missing = [
    'Color','Variety',
    'Processing.Method',
    'Region',
    'Country.of.Origin'
]

for col in categorical_missing:
    df_clean[col] = df_clean[col].fillna('Unknown')


# Numerical columns
numerical_missing = [
    'altitude_low_meters',
    'altitude_high_meters',
    'altitude_mean_meters',
    "Harvest.Year",
    'Quakers'
]

for col in numerical_missing:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())


# Check missing values again
missing = df_clean.isnull().sum()

print(missing[missing > 0])
```

```python
# final cleaning validation

print("Dataset shape:", df_clean.shape)

print("\nDuplicate rows:", df_clean.duplicated().sum())

print("\nMissing values:")
print(df_clean.isnull().sum().sum())

print("\nImportant data types:")
print(df_clean[
    [
        "Bag.Weight",
        "Harvest.Year",
        "Moisture",
        "Quakers",
        "altitude_low_meters",
        "altitude_high_meters",
        "altitude_mean_meters",
        "Total.Cup.Points"
    ]
].dtypes)

print("\nTarget summary:")
print(df_clean["Total.Cup.Points"].describe())
```

## Step 9 — Build the Preprocessing Pipeline

# Numerical → Scaling
- Number.of.Bags
- Harvest.Year
- Moisture
- Category.One.Defects
- Quakers
- Category.Two.Defects
- altitude_mean_meters
- altitude_range

# Categorical → Encoding
- Country.of.Origin
- Region
- In.Country.Partner
- Variety
- Processing.Method
- Color
- Certification.Body
- unit_of_measurement

StandardScaler → numerical features
OneHotEncoder → categorical features
ColumnTransformer → apply the correct transformation to each type

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

numerical_features=X_train.select_dtypes(
    include=['int64','float64']
).columns.tolist()

categorical_features=X_train.select_dtypes(
 include=['object']
).columns.tolist() 

preprocessor=ColumnTransformer(
    transformers=[
        (
            'num',StandardScaler(),numerical_features
        ),
        (
            'cat',OneHotEncoder(
                handle_unknown='ignore'
            ),
            categorical_features
        )
    ]
)
print('Numerical features:', numerical_features)
print('Categorical features:', categorical_features)
print('\nPreprocessor created successfully!')
```

```python
# Fit & Transform the Data

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print('X_train_processed shape:', X_train_processed.shape)
print('X_test_processed shape:', X_test_processed.shape)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

models = {
    'Linear Regression': y_pred_lr,
    'Decision Tree': y_pred_dt,
    'Random Forest': y_pred_rf,
    'Gradient Boosting': y_pred_gb,
    'XGBoost': y_pred_xgb
}

results = []

for name, predictions in models.items():

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            predictions
        )
    )

    r2 = r2_score(
        y_test,
        predictions
    )

    results.append([
        name,
        mae,
        rmse,
        r2
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        'Model',
        'MAE',
        'RMSE',
        'R² Score'
    ]
)

results_df = results_df.sort_values(
    'R² Score',
    ascending=False
)

print(results_df)
```

```python
# RandomizedSearchCV

from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor

rf=RandomForestRegressor(random_state=42,n_jobs=1)

random_search=RandomizedSearchCV(estimator=rf,param_distributions=param_grid,
                                n_iter=30,cv=5,scoring='r2',random_state=42,
                                n_jobs=-1,verbose=1)

random_search.fit(X_train_processed,y_train)

print('RandomizedSearchCV completed successfully')
```

```python
print('Best Parameters:')
print(random_search.best_params_)

print('\nBest Cross Validation R2:')
print(random_search.best_score_)
```

```python
# Evaluate Tuned Random Forest
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

best_rf = random_search.best_estimator_

y_pred_tuned = best_rf.predict(X_test_processed)

mae_tuned = mean_absolute_error(
    y_test,
    y_pred_tuned
)

rmse_tuned = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_tuned
    )
)

r2_tuned = r2_score(
    y_test,
    y_pred_tuned
)

print('Tuned Random Forest Performance')
print('MAE:', mae_tuned)
print('RMSE:', rmse_tuned)
print('R² Score:', r2_tuned)
```

```python
# Compare Original vs Tuned Model
comparison=pd.DataFrame({'Model':['Original Random Forest',
                                 'Tuned Random Forest'],
                        'MAE':[
                            mae_rf,mae_tuned],
                        'RMSE':[np.sqrt(mean_squared_error(y_test,y_pred_rf)),
                               rmse_tuned],
                        'R2 Score':[r2_score(y_test,y_pred_rf),
                                   r2_tuned]})
print(comparison)
```
